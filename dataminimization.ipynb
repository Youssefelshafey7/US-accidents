```python
import pandas as pd

path = "US_Accidents_March23.csv"
total_rows = 7_728_394
target = 1000
oversample = 1.5   # take a bit extra, trim later

cols = ["ID", "Severity", "Start_Time", "State", "Temperature(F)",
        "Visibility(mi)", "Weather_Condition", "Start_Lat", "Start_Lng"]  # add what you need

parts = []
frac = target * oversample / total_rows

for chunk in pd.read_csv(path, usecols=cols, chunksize=200_000):
    parts.append(
        chunk.groupby("Severity", group_keys=False)
             .apply(lambda g: g.sample(frac=frac, random_state=42))
    )

sample = pd.concat(parts, ignore_index=True)
print(len(sample))
```

```python
from sklearn.model_selection import train_test_split

sample, _ = train_test_split(
    sample, train_size=target, stratify=sample["Severity"], random_state=42
)
sample.to_csv("sample_1000.csv", index=False)
print(sample["Severity"].value_counts(normalize=True))
```
